# 00 · Inventario de trazabilidad del insumo

## Objetivo
Repetir el control de identidad binaria de BES-jun y dejar la ruta de cada copia junto a su hash completo.

## Alcance
Este cuaderno pertenece a la primera pasada de inventario. Ahora ejecuta el mismo control que `00_inventario_y_fuentes.ipynb`: tres rutas, tamaño, SHA-256 y MD5. No abre RB-1521, RB-IA-2191 ni el XLSM.


```python
from pathlib import Path
import os, sys
aqui = Path.cwd().resolve()
expediente = next(c for c in [aqui, *aqui.parents] if (c / "dataset.md").is_file() and (c / "entrada").is_dir())
os.chdir(expediente)
if str(expediente / "src") not in sys.path:
    sys.path.insert(0, str(expediente / "src"))
libro = expediente / "entrada" / "BES_jun.xlsx"

import hashlib
proyecto = next(c for c in expediente.parents if (c / "fuentes" / "originales").is_dir() and (c / "trabajo.md").is_file())
rutas = [
    libro,
    proyecto / "fuentes" / "originales" / "Copia de FINAL BES-jun.xlsx",
    proyecto / "Copia de FINAL BES-jun.xlsx",
]
registros = []
for ruta in rutas:
    datos = ruta.read_bytes()
    registros.append((str(ruta.relative_to(proyecto)), len(datos), hashlib.sha256(datos).hexdigest(), hashlib.md5(datos).hexdigest()))
for ruta, n, sha, md5 in registros:
    print(f"ruta={ruta}")
    print(f"bytes={n}")
    print(f"sha256={sha}")
    print(f"md5={md5}")
    print("---")
coinciden = len({r[1] for r in registros}) == 1 and len({r[2] for r in registros}) == 1
print("coincidencia_binaria=" + ("si" if coinciden else "no"))
assert coinciden
assert registros[0][2] == "ccb5a738baea13321993e8b6eccefd6cac90ecda86b1c4b9181f897fcfa83f85"
```

## Lectura del resultado
`coincidencia_binaria=si` significa que las tres rutas tienen el mismo tamaño y el mismo SHA-256. El hash no se abrevia en la salida: las tres líneas `sha256=` deben leerse completas y compararse entre sí.

El catálogo de fórmulas del proyecto está en `../../../docs/trazabilidad_ecuaciones.md`. Este cuaderno no lo calcula. La coincidencia de archivos no valida ese catálogo.
