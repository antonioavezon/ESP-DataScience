## F.1 Introducción científica

**Objetivo.** Separar el fenómeno físico de la tarea predictiva antes de comparar autores.

**Antecedentes.** El gas libre entra a una ESP cuando el líquido no lo disuelve por completo. El resumen del artículo de Zhu et al. (2019), página impresa 471, dice que el incremento de presión se degrada al aumentar la fracción volumétrica de gas en la admisión, y que el *surging* y las inestabilidades dependen de esa fracción y de la velocidad.

**Explicación conceptual.** Conviene fijar las palabras:

- Incremento de presión: diferencia de presión entre dos puntos de la bomba. Aquí, `dp2_3_psi` es el incremento de la etapa 3, en libras por pulgada cuadrada (psi), con certeza media. No es la presión absoluta, ni la altura de columna, ni la presión de fondo.
- Caudal líquido, Ql: volumen de líquido por unidad de tiempo. En las tablas, `ql_bpd` está en barriles por día (bpd).
- Caudal de gas, Qg: volumen de gas por unidad de tiempo, en las condiciones en que se defina.
- Fracción volumétrica de gas (GVF, *gas volumetric fraction*): Qg / (Qg + Ql), ecuación (2.5) de la tesis, página impresa 53.
- Caudal de gas normalizado, Qgd: Qg / Qmax, ecuación (2.7) de la misma página. Qmax es el caudal abierto de líquido. Qgd es adimensional: las unidades de Qg y de Qmax se cancelan si son las mismas.
- Patrón de flujo: modo en que se organizan gas y líquido. El artículo de 2019 nombra cuatro: burbuja dispersa, burbuja, intermitente y segregado.
- Ensayo de surging, en ese artículo: el caudal de líquido queda fijo y el de gas aumenta. No es un sinónimo de cualquier inestabilidad de bombeo. El fenómeno llamado *pressure surging* es una inestabilidad descrita por los autores; el ensayo es el procedimiento que la recorre.
- Ensayo de mapping: el caudal de gas queda fijo y el de líquido varía.
- Capacidad predictiva: cuánto se acerca un número calculado a `dp2_3_psi` medido, por ejemplo con el error absoluto medio (MAE, *mean absolute error*), en psi.
- Interpretación física: una lectura del mecanismo. Puede coincidir con una tendencia y, aun así, no ser una medición del régimen.

**Evidencia utilizada.** Resumen de 2019 y ecuaciones (2.5) y (2.7) de la tesis. Las cifras de M1–M5 no se recalculan aquí.

**Interpretación.** Predecir psi y explicar un patrón son preguntas distintas. Pueden complementarse.

**Limitaciones.** No se releyó la tesis entera ni el inventario de correlaciones de la revisión de 2018.

**Conexión.** La sección siguiente separa cada documento de Zhu.

## F.4 Comparabilidad

**Objetivo.** Mostrar, con un ejemplo que no es de la bomba, por qué Qgd y GVF pueden dar números distintos, y por qué un MAE en psi no se compara con un acierto de patrón.

**Antecedentes.** La tesis, página impresa 53, escribe:

$$\mathrm{GVF} = \frac{Q_g}{Q_g + Q_l} \qquad (2.5)$$

$$Q_{gd} = \frac{Q_g}{Q_{\max}} \qquad (2.7)$$

Qg es el caudal volumétrico de gas en la admisión. Ql es el caudal volumétrico de líquido en la admisión. Qmax es el caudal abierto de líquido, la referencia de esa definición. En surging, `gvf0_porcentaje` es un porcentaje de GVF, no el código Qgd de mapping.

**Explicación conceptual.** Una magnitud adimensional es un cociente en el que las unidades se cancelan. Qgd no trae unidad de caudal. Para pasar de Qgd a Qg haría falta el Qmax de esa velocidad, medido, no supuesto. La tesis publica 4900 bpd a 3500 rpm. No se aplica esa cifra a las filas, y no se inventa el Qmax de 1800 rpm.

Un error de presión en psi mide la distancia entre dos incrementos. Un clasificador de patrones acierta o falla una etiqueta. No tienen la misma unidad ni el mismo objetivo. Por eso no hay una tabla de MAE entre M1–M5 y el modelo de fronteras.

La celda siguiente usa caudales ficticios, en unidades arbitrarias e iguales. No entran en los resultados de la TE-2700.

**Evidencia utilizada.** Ecuaciones (2.5) y (2.7). El dossier de código cita el informe y mantiene cerrada la puerta numérica. Un booleano verdadero no se acepta como verificación.

**Interpretación.** Dos fórmulas pueden coincidir en un caso elegido y diferir en otro. La coincidencia no las vuelve la misma variable.

**Limitaciones.** El ejemplo no calibra nada. La puerta administrativa, si alguna vez se abriera con registros citados, seguiría dejando la equivalencia física en falso.

**Conexión.** El gráfico de F.5 muestra mediciones del CSV, no una conversión de Qgd a bpd.


```python
print("EJEMPLO DIDÁCTICO FICTICIO. No son caudales de la bomba.")
def mostrar(qg, ql, qmax):
    gvf = qg / (qg + ql)
    qgd = qg / qmax
    print(f"Qg={qg}, Ql={ql}, Qmax={qmax} -> GVF={gvf:.3f}, Qgd={qgd:.3f}")

mostrar(100, 900, 1000)
mostrar(100, 400, 1000)

filas = []
for registro in dossier_de_la_actividad_215():
    filas.append({
        "condición": registro.condicion,
        "resultado": registro.resultado,
        "estado": registro.estado,
        "fuente": registro.fuente,
    })
try:
    exigir_comparacion_numerica(dossier_de_la_actividad_215())
    raise RuntimeError("La puerta no debía abrirse.")
except ValueError as exc:
    print(exc)
try:
    exigir_comparacion_numerica(True)
    raise RuntimeError("Un booleano no debía aceptarse.")
except TypeError as exc:
    print(exc)
pd.DataFrame(filas)
```
